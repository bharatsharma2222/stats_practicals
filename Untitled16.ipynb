```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
import statsmodels.stats.anova as anova
```

```python
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/AER/TeachingRatings.csv"
```

```python
df = pd.read_csv(url)
```


Q1. Regression with T-Test

Using the Teachers' Rating dataset, does gender affect teaching evaluation rates?

The output should contain:

OLS Regression Results
Dependent variable: eval
Independent variable: female
Coefficient (coef)
t-value (t)
p-value (P>|t|)
R-squared
F-statistic
---



```python
print("First 5 rows of dataset:")
print(df.head())
```

```python
print("\nDataset shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)
```

```python

df["female"] = (df["gender"] == "female").astype(int)
```

```python

y = df["eval"]
X = df["female"]
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print("\nOLS Regression Results:")
print(model.summary())
```

```python

print("\nT-Test for Gender:")
print("t-value =", model.tvalues["female"])
print("p-value =", model.pvalues["female"])
```

```python
if model.pvalues["female"] < 0.05:
    print("\nConclusion:")
    print("Gender has a statistically significant effect on teaching evaluation.")
else:
    print("\nConclusion:")
    print("Gender does not have a statistically significant effect on teaching evaluation.")
```

Q2. Regression with ANOVA

Using the Teachers' Rating dataset, does the beauty score for instructors differ by age?

The output should be:

	df	sum_sq	mean_sq	F	PR(>F)
age_group	2.0	20.422744	10.211372	17.597559	4.322549e-08

```python
print("\n")
print("=" * 60)
print("Q2. REGRESSION WITH ANOVA")
print("=" * 60)
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 35, 50, 100],
    labels=["Young", "Middle", "Old"]
)
```

```python
model2 = ols(
    "beauty ~ C(age_group)",
    data=df
).fit()
```

```python
anova_table = anova.anova_lm(model2)
anova_table.index = ["age_group", "Residual"]

print("\nANOVA Results:")
print(anova_table)
```

```python

p_value = anova_table.loc["age_group", "PR(>F)"]

print("\nP-value:", p_value)

if p_value < 0.05:
    print("\nConclusion:")
    print("Beauty score differs significantly by age group.")
else:
    print("\nConclusion:")
    print("Beauty score does not differ significantly by age group.")
```

```python
X = df["beauty"]
X = sm.add_constant(X)
y = df["eval"]
model3 = sm.OLS(y, X).fit()
print(model3.summary())
```

```python
correlation = df["eval"].corr(df["beauty"])

print("\nPearson Correlation:")
print("Correlation =", correlation)

# Conclusion
if model3.pvalues["beauty"] < 0.05:
    print("\nConclusion:")
    print("Teaching evaluation is significantly correlated with beauty score.")
else:
    print("\nConclusion:")
    print("Teaching evaluation is not significantly correlated with beauty score.")
```
