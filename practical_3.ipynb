```python
# ============================================================
# STUDENT / TEACHERS RATING DATASET
# Q1 TO Q5
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/AER/TeachingRatings.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully!")
print(df.head())
print("\nShape:", df.shape)


# ============================================================
# Q1. Percentage of visible minorities who are tenure professors
# ============================================================

print("\n========== Q1 ==========")

# Visible minority professors
minority = df[df["minority"] == "yes"]

# Visible minority + tenured
minority_tenure = minority[minority["tenure"] == "yes"]

# Percentage
percentage = (len(minority_tenure) / len(minority)) * 100

print("Percentage of visible minorities who are tenure professors:",
      round(percentage, 2), "%")

# Tenure status by minority
table = pd.crosstab(
    df["minority"],
    df["tenure"],
    normalize="index"
) * 100

print("\nTenure status by minority:")
print(table)

print("\nConclusion:")
print("Compare the percentages to see whether tenure status differs by")
print("visible minority status.")


# ============================================================
# Q2. Does average age differ by tenure?
# ============================================================

print("\n========== Q2 ==========")

# Tenured professors
tenured = df[df["tenure"] == "yes"]["age"]

# Untenured professors
untenured = df[df["tenure"] == "no"]["age"]

print("Tenured Professors")
print("Mean =", round(tenured.mean(), 2))
print("Standard Deviation =", round(tenured.std(), 2))

print("\nUntenured Professors")
print("Mean =", round(untenured.mean(), 2))
print("Standard Deviation =", round(untenured.std(), 2))


# ============================================================
# Q3. Which graph works better for age?
# ============================================================

print("\n========== Q3 ==========")

print("Histogram works better for the age variable because age is numerical.")

plt.hist(df["age"], bins=10)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution of Professors")
plt.show()


# ============================================================
# Q4. Difference between bar and barh
# ============================================================

print("\n========== Q4 ==========")

print("plt.bar()  = Vertical bar graph")
print("plt.barh() = Horizontal bar graph")

# Gender count
gender_count = df["gender"].value_counts()

# Bar graph
plt.bar(gender_count.index, gender_count.values)
plt.xlabel("Gender")
plt.ylabel("Number of Professors")
plt.title("Gender Distribution")
plt.show()


# ============================================================
# Q5. Median evaluation score for tenured professors
# ============================================================

print("\n========== Q5 ==========")

tenured_eval = df[df["tenure"] == "yes"]["eval"]

median_eval = tenured_eval.median()

print("Median evaluation score for tenured professors:",
      median_eval)
```
