 Q2. Find the mean, median, minimum and maximum values for the students

```python
print(" STUDENTS STATISTICS")

students_mean = df["students"].mean()
students_median = df["students"].median()
students_min = df["students"].min()
students_max = df["students"].max()

print("Mean    :", students_mean)
print("Median  :", students_median)
print("Minimum :", students_min)
print("Maximum :", students_max)
```

Q5. Does average beauty score differ by gender? Produce the means and standard deviations for both male and female instructors

```python
print(" BEAUTY SCORE BY GENDER")

gender_beauty = df.groupby("gender")["beauty"].agg(
    Mean="mean",
    Standard_Deviation="std",
    Count="count"
)

print(gender_beauty)

male_mean = df[df["gender"] == "male"]["beauty"].mean()
male_sd = df[df["gender"] == "male"]["beauty"].std()

female_mean = df[df["gender"] == "female"]["beauty"].mean()
female_sd = df[df["gender"] == "female"]["beauty"].std()

print("\nMale:")
print("Mean =", male_mean)
print("SD   =", male_sd)

print("\nFemale:")
print("Mean =", female_mean)
print("SD   =", female_sd)
```
