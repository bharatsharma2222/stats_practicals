**Q1. Regression with T-test,
 Does gender affect teaching evaluation rates?**

```python
!pip install statsmodels pandas -q

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.formula.api import ols
from scipy.stats import pearsonr

# Load TeachingRatings dataset
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ST0151EN-SkillsNetwork/labs/teachingratings.csv"

df = pd.read_csv(url)

# Display dataset
print(df.head())

# Dataset information
print(df.shape)
print(df.describe())
```

```python
# Convert gender into numerical values
df['female'] = (df['gender'] == 'female').astype(int)

# Regression model
model1 = smf.ols('eval ~ female', data=df).fit()

# Display regression results
print(model1.summary())
```

**Conclusion: The regression analysis shows that gender has a statistically significant effect on teaching evaluation scores. The coefficient for male instructors is 0.168, with a p-value of 0.001, which is less than 0.05. This indicates that male instructors received, on average, approximately 0.168 higher evaluation scores than female instructors. However, the R-squared value of 0.022 shows that gender explains only 2.2% of the variation in evaluation scores.**

```python
from scipy.stats import ttest_ind

# Separate evaluation scores by gender
male = df[df["gender"] == "male"]["eval"]
female = df[df["gender"] == "female"]["eval"]

# Independent samples t-test
t_stat, p_value = ttest_ind(male, female, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject H0: Gender significantly affects teaching evaluation rates.")
else:
    print("Fail to reject H0: Gender does not significantly affect teaching evaluation rates.")
```

**Q2. Regression with ANOVA**

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 40, 50, 100],
    labels=["Young", "Middle", "Old"]
)

# Check age groups
print(df["age_group"].value_counts())

# Regression model
model = smf.ols("beauty ~ C(age_group)", data=df).fit()

# ANOVA table
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```

**Conclusion: The ANOVA test produced an F-statistic of 15.189 and a p-value of 4.094 × 10⁻⁷, which is less than 0.05. Therefore, we reject the null hypothesis and conclude that beauty scores differ significantly among the Young, Middle, and Old age groups. This indicates that age group has a statistically significant relationship with beauty scores.**

```python
anova_table.index = ["age_group" if "C(age_group)" in str(x) else x
                     for x in anova_table.index]

print(anova_table)
```

```python
p_value = anova_table.iloc[0]["PR(>F)"]

if p_value < 0.05:
    print("Reject H0: Beauty score differs significantly by age group.")
else:
    print("Fail to reject H0: Beauty score does not differ significantly by age group.")
```

**Conclusion: The ANOVA test gives an F-statistic of 15.189 and a p-value of 0.000000409, which is less than 0.05. Therefore, we reject the null hypothesis (H₀) and conclude that beauty scores differ significantly among the Young, Middle, and Old age groups. This indicates a statistically significant association between age group and beauty scores.**

##**Question 3: Is teaching evaluation score correlated with beauty score?**

```python
from scipy.stats import pearsonr
# Pearson correlation
correlation, p_value = pearsonr(df["eval"], df["beauty"])

print("Correlation coefficient:", correlation)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("There is a statistically significant correlation between evaluation and beauty.")
else:
    print("There is no statistically significant correlation between evaluation and beauty.")
```

```python
# Regression model
model3 = smf.ols('eval ~ beauty', data=df).fit()

# Display regression results
print(model3.summary())
```

**The Pearson correlation coefficient is 0.189, indicating a weak positive correlation between teaching evaluation scores and beauty scores. The p-value is 0.0000425, which is less than 0.05, so the correlation is statistically significant. Therefore, there is evidence of a positive relationship between teaching evaluation scores and beauty scores, although the relationship is weak.**
