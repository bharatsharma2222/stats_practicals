```python
import pandas as pd
import math
from scipy.stats import norm
```

```python
data = pd.read_csv("/content/evalsdataset.csv")
```

```python
scores = data["score"]
```

##**Q1. Using the teachers' rating dataset, what is the probability of receiving an evaluation score of greater than 4.5**

```python
total = len(scores)
greater_4_5 = (scores > 4.5).sum()

prob_q1 = greater_4_5 / total

print("Q1: Probability of score > 4.5")
print("Number of scores > 4.5:", greater_4_5)
print("Total scores:", total)
print("Probability:", prob_q1)
print("Percentage:", prob_q1 * 100, "%")
```

##**Q2. Using the teachers' rating dataset, what is the probability of receiving an evaluation score greater than 3.5 and less than 4.2**

```python
between = ((scores > 3.5) & (scores < 4.2)).sum()

prob_q2 = between / total

print("\nQ2: Probability of 3.5 < score < 4.2")
print("Number of scores:", between)
print("Total scores:", total)
print("Probability:", prob_q2)
print("Percentage:", prob_q2 * 100, "%")
```

###**Q3. Using the two-tailed test from a normal distribution: 1. A professional basketball team wants to compare its performance with that of players in a regional league. 2. The pros are known to have a historic mean of 12 points per game with a standard deviation of 5.5. 3. A group of 36 regional players recorded on average 10.7 points per game. 4. The pro coach would like to know whether his professional team scores on average are different from that of the regional players. State the null hypothesis Null Hypo: ("The mean point of the regional players is not different from the historic mean") Alternative Hypo: ("The mean point of the regional players is different from the historic mean")**

```python
# Part 1-4:
historic_mean = 12
standard_deviation = 5.5
sample_size = 36
sample_mean = 10.7
```

```python
print("Historic Mean =", historic_mean)
print("Standard Deviation =", standard_deviation)
print("Sample Size =", sample_size)
print("Sample Mean =", sample_mean)
```

```python
# Null Hypothesis (H0)
print("\nNull Hypothesis (H0):")
print("The mean point of the regional players is not different from the historic mean.")

# Alternative Hypothesis (H1)
print("\nAlternative Hypothesis (H1):")
print("The mean point of the regional players is different from the historic mean.")
```

```python
#Z-score

z = (sample_mean - historic_mean) / (standard_deviation / math.sqrt(sample_size))

print("\nZ-score =", z)
```

```python
#find P-value
from scipy.stats import norm

p_value = 2 * norm.cdf(-abs(z))

print("P-value =", p_value)
```

```python
#at 5% significance level
alpha = 0.05
critical_value = 1.96

if abs(z) > critical_value:
    print("\nReject H0")
else:
    print("\nFail to reject H0")
```

```python
#conclusion
if p_value < alpha:
    print("Conclusion: The mean points scored by regional players is significantly different from 12.")
else:
    print("Conclusion: There is not enough evidence to say that the mean points scored by regional players is different from 12.")
```

##**Conclusion:  There is not enough evidence to say that the mean points scored by regional players is different from 12.**
