```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/AER/TeachingRatings.csv"

df = pd.read_csv(url)

print("First 5 observations:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)


# ============================================================
# QUESTION 1
# Identify duplicate cases using prof.
# Find mean and standard deviation of age using all observations.
# Repeat after keeping one observation per instructor (94 observations).
# ============================================================

print("\n================ QUESTION 1 ================")

# Find duplicate professors
duplicates = df[df.duplicated("prof", keep=False)]

print("\nDuplicate cases using prof:")
print(duplicates[["prof", "age"]].sort_values("prof"))

# Average age using ALL observations
print("\nUsing ALL observations:")
print("Average age =", df["age"].mean())
print("Standard deviation =", df["age"].std())

# Keep only one observation for each professor
df_unique = df.drop_duplicates(subset="prof", keep="first")

print("\nAfter removing duplicate professors:")
print("Number of observations =", len(df_unique))

print("Average age =", df_unique["age"].mean())
print("Standard deviation =", df_unique["age"].std())


# ============================================================
# QUESTION 2
# Using a bar chart, demonstrate if instructors teaching
# lower-division courses receive higher average evaluations.
# ============================================================

print("\n================ QUESTION 2 ================")

# Average evaluation by division
division_avg = df.groupby("division")["eval"].mean()

print("\nAverage teaching evaluation by division:")
print(division_avg)

# Bar chart
plt.figure(figsize=(6, 4))
plt.bar(division_avg.index, division_avg.values)

plt.xlabel("Course Division")
plt.ylabel("Average Teaching Evaluation")
plt.title("Average Teaching Evaluation by Course Division")

plt.show()


# ============================================================
# QUESTION 3
# Plot relationship between age and teaching evaluation scores.
# ============================================================

print("\n================ QUESTION 3 ================")

plt.figure(figsize=(7, 5))

plt.scatter(df["age"], df["eval"])

plt.xlabel("Age")
plt.ylabel("Teaching Evaluation Score")
plt.title("Age vs Teaching Evaluation")

plt.show()


# ============================================================
# QUESTION 4
# Gender-differentiated scatter plots:
# Age vs Teaching Evaluation
# ============================================================

print("\n================ QUESTION 4 ================")

plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x="age",
    y="eval",
    hue="gender"
)

plt.xlabel("Age")
plt.ylabel("Teaching Evaluation Score")
plt.title("Age vs Teaching Evaluation by Gender")

plt.show()


# ============================================================
# QUESTION 5
# Scatterplot of age and evaluation scores,
# differentiated by gender AND tenure.
# ============================================================

print("\n================ QUESTION 5 ================")

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="age",
    y="eval",
    hue="gender",
    style="tenure",
    s=80
)

plt.xlabel("Age")
plt.ylabel("Teaching Evaluation Score")
plt.title("Age vs Evaluation by Gender and Tenure")

plt.show()
```
