```python
import pandas  as pd
```

```python
import numpy as np

data = {
    "name" : ["Aarush","ayush","Ansh","Tushar", "anshul"],
    "mango db" : [58,81,25,79,67],
    "software engineering" :[78,98,63,54,78],
    "C++" : [82,87,75,45,65],
    "python" : [95,19,64,83,92],
}

df = pd.DataFrame(data)

print("dataframe:")
print(df)
```

```python
A = df[["mango db","software engineering","C++","python"]].to_numpy()
print("numpy array:")
print(A)

print("shape:",A.shape)
print("size:",A.size)
print("dimension:",A.ndim)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
```

```python
addition = A + B
print("Addition:")
print(addition)
```

```python
subtraction = A - B
print("Subtraction:")
print(subtraction)
```

```python
multiplication = A * B
print("Element-wise Multiplication:")
print(multiplication)
```

```python
matrix_product = A @ B
print("Matrix Multiplication:")
print(matrix_product)
```

```python
transpose = A.T
print("Transpose:")
print(transpose)
```

```python
inverse = np.linalg.inv(A)
print("Inverse:")
print(inverse)
```

```python
print("A × A_inverse:")
print(A @ inverse)
```

```python
product = np.dot(A, B)
print("Product:")
print(product)

```
